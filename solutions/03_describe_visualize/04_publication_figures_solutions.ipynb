## Learning goals

- **A**pply a consistent, accessible style to empirical figures.
- **D**istinguish uncertainty intervals from point estimates.
- **E**xport vector and raster graphics reproducibly.

```python
from io import BytesIO

import matplotlib.pyplot as plt
import numpy as np

effects = np.array([-0.12, 0.08, 0.22])
standard_errors = np.array([0.05, 0.06, 0.09])
labels = ["Control A", "Control B", "Treatment"]
print("Fictional estimates:", effects)
```

## Why this matters

A journal or report figure should make the question, units, and uncertainty clear without relying on color alone. The regression estimates here are invented to teach plotting—not outcomes from an empirical study.

## Create a consistent style

Choose readable font sizes, transparent grid lines, and one colorblind-safe color. Use a style context so changes do not leak into later notebooks.

```python
house_style = {"axes.spines.top": False, "axes.spines.right": False,
               "axes.grid": True, "grid.alpha": 0.2,
               "font.size": 11, "figure.dpi": 120}
with plt.rc_context(house_style):
    fig, ax = plt.subplots(figsize=(6, 4))
    ax.plot([2022, 2023, 2024], [100, 104, 110],
            marker="o", color="#0072B2", linewidth=2)
    ax.set(xlabel="Year", ylabel="Illustrative index (base 100)")
    fig.tight_layout()
plt.close(fig)
print("Consistent line chart created")
```

## Plot estimates and confidence intervals

A point estimate is a model result. A conventional illustrative interval is estimate plus or minus 1.96 standard errors; actual inference depends on the model assumptions and reference distribution.

```python
ci_low = effects - 1.96 * standard_errors
ci_high = effects + 1.96 * standard_errors
fig_coef, ax_coef = plt.subplots(figsize=(6, 4))
ax_coef.errorbar(effects, range(len(labels)),
                 xerr=1.96 * standard_errors, fmt="o", capsize=3,
                 color="#0072B2")
ax_coef.set_yticks(range(len(labels)), labels=labels)
ax_coef.axvline(0, color="black", linestyle="--")
ax_coef.set_xlabel("Illustrative coefficient (unit change)")
fig_coef.tight_layout()
plt.close(fig_coef)
```

## Save raster and vector versions

PNG is useful for slides; PDF and SVG preserve vector lines. Keep a descriptive caption alongside each figure and regenerate it from code when inputs change.

```python
png = BytesIO()
pdf = BytesIO()
fig_coef.savefig(png, format="png", dpi=300)
fig_coef.savefig(pdf, format="pdf")
print("PNG bytes:", len(png.getvalue()))
print("PDF bytes:", len(pdf.getvalue()))
```

## Common errors

- Using an interval formula without checking degrees of freedom or the model's standard-error assumptions can misstate uncertainty.
- A color-only legend may be unreadable for some viewers; use markers and direct labels.
- Raster files at low resolution can blur in print; use a vector export when practical.

## Recap

- Style settings make charts consistent.
- Labels and uncertainty intervals are part of scientific interpretation.
- Export figures directly from reproducible source code.
