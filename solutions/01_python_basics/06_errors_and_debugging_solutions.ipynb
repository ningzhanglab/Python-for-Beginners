## Why this matters

Real surveys include missing IDs and numbers stored as strings. Python's **traceback** shows where an error occurred and which exception was raised. Rather than deleting failing lines, turn the error into a testable, understandable case.

## Reproduce a KeyError safely

The following snippet deliberately triggers a missing dictionary key, catches only the expected error, and prints a short message. Broad `except:` clauses can hide real bugs.

```python
survey_income = {'Ana': '42,000', 'Ben': '50,000'}
try:
    print(survey_income['Missing'])
except KeyError as error:
    print('Missing key:', error)
```

## Read error types

A typo may cause `NameError`, a bad index may cause `IndexError`, and mixing types may cause `TypeError`. Each means something different.

```python
values = [10, 20]
try:
    print(values[3])
except IndexError as error:
    print('Index problem:', type(error).__name__)
print('Known value:', values[0])
```

## Make a minimal reproducible example

A small sample and one clear assertion are easier to debug than an entire data pipeline. Convert a numeric string only after removing presentation characters.

```python
wage_text = survey_income['Ana']
wage_number = int(wage_text.replace(',', ''))
assert wage_number > 0
print('Parsed wage:', wage_number)
```

## Guard against invalid denominators

When a baseline is zero, a growth-rate formula is undefined. Check it explicitly and return a clear missing value instead of hiding the problem.

```python
def safe_growth_rate(first, last):
    if first == 0:
        return None
    return (last / first - 1) * 100

print('Growth with zero baseline:', safe_growth_rate(0, 2))
```

### ★ Exercise 1 — Safe dictionary lookup

Define `lookup_income(name)` to return the survey's string amount or `None` for an unknown name.

```python
def lookup_income(name):
    return survey_income.get(name)
```

```python
if lookup_income('Ana') == survey_income['Ana'] and lookup_income('Unknown') is None:
    print('✓ Safe lookup works')
else:
    print('Hint: dictionary .get(key) returns None when a key is absent.')
```

```python
parsed_wage = int('50,000'.replace(',', ''))
```

```python
if isinstance(parsed_wage, int) and parsed_wage == 50000:
    print('✓ Wage parsed')
else:
    print('Hint: remove the comma before converting the text to int.')
```

### ★★★ Exercise 3 — Safe percentage change

Define `safe_percent_change(old, new)` to return `None` if `old` is zero, or the percentage change otherwise.

```python
def safe_percent_change(old, new):
    if old == 0:
        return None
    return (new / old - 1) * 100
```

```python
test_change = safe_percent_change(100, 105)
if (safe_percent_change(0, 10) is None and
    isinstance(test_change, (int, float)) and math.isclose(test_change, 5)):
    print('✓ Safe growth function works')
else:
    print('Hint: check the denominator before doing the division.')
```

## Common errors

- `NameError`: a variable was never defined (or a notebook cell was skipped).
- `KeyError`: a requested dictionary key does not exist.
- `IndexError`: a list position is outside the valid range.
- `TypeError` or `ValueError`: a numeric conversion or operation received an unexpected type or value.
