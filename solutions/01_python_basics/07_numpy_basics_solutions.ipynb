## Why this matters

Economists work with many observations at once. A NumPy **array** stores numbers in a regular structure and supports vectorized arithmetic, meaning operations apply across elements. Matrix multiplication is a bridge to ordinary least squares (OLS) later in the course.

## Arrays and elementwise operations

In NumPy, dividing two equally shaped arrays divides matching positions. This is different from dividing Python lists.

```python
prices = np.array([100.0, 110.0, 121.0])
growth = prices[1:] / prices[:-1] - 1
print('Growth rates:', growth)
print('Average price:', prices.mean())
```

## Reproducible random simulation

Use `np.random.default_rng(seed)` rather than resetting global random state. The same seed gives the same pseudo-random sequence in the same NumPy implementation.

```python
rng = np.random.default_rng(123)
shock = rng.normal(loc=0, scale=1, size=4)
print('Simulated shocks:', shock)
print('Shock average:', shock.mean())
```

## Matrix operations: the OLS bridge

The `@` operator performs matrix multiplication; `.T` transposes an array. Here we solve a tiny regression with an intercept and one predictor. Solving the linear system is usually preferable to explicitly computing an inverse.

```python
x = np.array([[1.0, 1.0], [1.0, 2.0], [1.0, 3.0]])
y = np.array([2.0, 4.0, 6.0])
coefficients = np.linalg.solve(x.T @ x, x.T @ y)
print('Intercept and slope:', coefficients)
assert np.allclose(coefficients, [0.0, 2.0])
```

## Common errors

- `TypeError`: dividing Python lists is not the same as dividing numeric NumPy arrays.
- `ValueError: operands could not be broadcast together`: elementwise array shapes are incompatible.
- `LinAlgError: Singular matrix`: the matrix system lacks a unique solution.
- Unseeded randomness changes on each run; use `default_rng` with a documented seed.

## Recap

- NumPy arrays provide vectorized numeric operations.
- Seeded generators make classroom simulations reproducible.
- Matrix operations underpin later regression methods.
