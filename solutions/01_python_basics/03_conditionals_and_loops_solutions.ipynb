## Why this matters

An empirical dataset often needs categories such as low, middle, and high income. A conditional branch selects what to do; loops repeat a step over observations. These tools are especially useful for small checks and automation before introducing pandas vectorization.

## if / elif / else

Use comparisons to assign a category. Here boundaries are illustrative, not a policy or scientific income threshold.

```python
income = 45000
if income < 30000:
    category = 'low'
elif income < 70000:
    category = 'middle'
else:
    category = 'high'
print('Income bracket:', category)
```

## Loop, range, and enumerate

A `for` loop repeats actions. `enumerate` supplies both an index and a value, and `range` generates integer positions.

```python
incomes = [20000, 45000, 80000]
for index, value in enumerate(incomes):
    print(index, value)
for year in range(2023, 2025):
    print('Survey wave:', year)
```

## zip and a list comprehension

`zip` pairs values in order. A list comprehension creates a new list from another sequence in a compact, readable form.

```python
names = ['Ana', 'Ben', 'Chao']
wage_by_person = dict(zip(names, incomes))
brackets = ['low' if wage < 30000 else 'middle' if wage < 70000
            else 'high' for wage in incomes]
print(wage_by_person)
print(brackets)
```

## while loops need stopping rules

A `while` loop continues as long as its condition stays true. Always update the value that controls it to avoid an infinite loop.

```python
period = 0
while period < 3:
    period += 1
print('Completed periods:', period)
```

### ★ Exercise 1 — Classify an income

Set `income_label` to `low`, `middle`, or `high` for `income_to_classify = 73000`, using the boundaries above.

```python
income_to_classify = 73000
income_label = ('low' if income_to_classify < 30000 else
                'middle' if income_to_classify < 70000 else 'high')
```

```python
expected_label = ('low' if income_to_classify < 30000 else
                  'middle' if income_to_classify < 70000 else 'high')
if income_label == expected_label:
    print('✓ Income category is correct')
else:
    print('Hint: compare the value against the lower and upper boundaries.')
```

### ★★ Exercise 2 — Total wages with a loop

Compute `total_income` as the sum of `incomes`. Try a `for` loop instead of typing the total.

```python
total_income = 0
for amount in incomes:
    total_income += amount
```

```python
if isinstance(total_income, (int, float)) and total_income == sum(incomes):
    print('✓ Loop sum is correct')
else:
    print('Hint: start an accumulator at zero, then add each amount.')
```

```python
if isinstance(survey_lookup, dict) and survey_lookup == dict(zip(names, incomes)):
    print('✓ Survey lookup is correct')
else:
    print('Hint: zip combines two aligned lists into pairs.')
```

### ★★★ Exercise 4 — Filter above-threshold wages

Set `high_incomes` to a list of entries in `incomes` that exceed 40000.

```python
high_incomes = [amount for amount in incomes if amount > 40000]
```

```python
expected_high = [value for value in incomes if value > 40000]
if isinstance(high_incomes, list) and high_incomes == expected_high:
    print('✓ Filtered list is correct')
else:
    print('Hint: use a list comprehension with an if condition.')
```

## Common errors

- `IndentationError`: code inside a loop or condition needs consistent indentation.
- An infinite `while` loop happens if its condition never changes; update the counter each iteration.
- `TypeError` comparing text to numbers means you must convert the input to a numeric type first.

## Recap

- `if/elif/else` picks actions based on conditions.
- `for` iterates over a collection, and `while` repeats until a condition changes.
- `zip`, `enumerate`, and comprehensions automate small research tasks.
