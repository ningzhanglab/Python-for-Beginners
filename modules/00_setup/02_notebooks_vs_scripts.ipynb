## Why this matters

A research notebook is useful for exploring ideas, but its kernel remembers variables from previously executed cells. A script is a plain `.py` file that can be run from a fresh process. Both can be reproducible when inputs and execution order are explicit.

## Cell order and hidden state

Cells in one notebook share a live **kernel** (the running Python process). Run the following cells in order, then restart the kernel and run everything again. In an analysis, changing an earlier input means rerunning dependent cells.

```python
prices = [80, 84, 90]
changes = [(new / old - 1) * 100 for old, new in zip(prices, prices[1:])]
print('Prices:', prices)
print('Growth rates:', changes)
```

## Turn a calculation into a script

`runpy.run_path` executes a Python file in a fresh namespace. This example uses a temporary folder, so it does not write new files to your course checkout.

```python
with tempfile.TemporaryDirectory() as folder:
    path = Path(folder) / 'growth_script.py'
    path.write_text('start, end = 100, 110\n'
                    'growth = (end / start - 1) * 100\n', encoding='utf-8')
    script_result = runpy.run_path(str(path))
    print('Script growth:', script_result['growth'])
```

## Check results rather than trusting cell order

`assert` can catch wrong answers in a research pipeline. When checking decimal arithmetic, use `math.isclose` to avoid floating-point surprises.

```python
assert len(changes) == len(prices) - 1
assert math.isclose(changes[0], 5.0)
print('Research calculation checks passed.')
```

## If you know Stata/R

| Research action | Python | Stata | R |
| --- | --- | --- | --- |
| Store a value | `wage = 100` | `local wage = 100` | `wage <- 100` |
| Run saved code | `python analysis.py` | `do analysis.do` | `Rscript analysis.R` |
| Check a result | `assert total > 0` | `assert total > 0` | `stopifnot(total > 0)` |

For a notebook, **Restart Kernel and Run All** is a good hidden-state check.

## Common errors

- `NameError`: a name was created in a cell you did not run after a restart. Run the notebook top to bottom.
- `FileNotFoundError`: your script or data path depends on the working directory. Use `pathlib` and explicit paths.
- Unexpected old results: a previous kernel retains variables. Restart the kernel, then run all cells.

## Recap

- A notebook kernel keeps state across cells.
- Scripts package reusable analysis code in `.py` files.
- A clean restart and assertions make research more reproducible.
