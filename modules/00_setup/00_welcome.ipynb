## Why notebooks?

An empirical researcher often starts by asking a small question about data. A **notebook** lets you keep the question, code, and results together. A **cell** is one block of runnable code or written explanation.

## A first calculation

Suppose a basket of goods cost 100 currency units last year and 104 this year. We can calculate its percentage price increase.

```python
price_last_year = 100
price_this_year = 104
price_change_pct = (price_this_year / price_last_year - 1) * 100
print(f"Price increase: {price_change_pct:.1f}%")
```

The line above creates variables, performs arithmetic, and prints a readable result. You will learn the details in Module 1.

Run a cell using **Shift+Enter**. In Jupyter/Colab, an execution count appears near the cell after it runs.

```python
research_question = "How fast have prices changed?"
print(research_question)
print(type(price_change_pct).__name__)
```

## Common errors

- `NameError: name 'gdp' is not defined` — you used a name before assigning it. Run its definition cell first.
- `TypeError: unsupported operand type(s) for /: 'str' and 'int'` — one input is text, not a number. Check the value's type.
- Out-of-order results — notebooks retain variables from previously run cells. Choose **Runtime → Restart session** (Colab) or **Kernel → Restart Kernel** (Jupyter), then run everything top to bottom.
