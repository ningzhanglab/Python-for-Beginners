## Learning goals

- **Derive** a simple OLS intercept and slope with NumPy.
- **Check** residuals, fitted values, and R-squared.
- **Explain** why an estimated association is not automatically causal.

```python
import numpy as np

experience = np.arange(1, 9, dtype=float)
noise = np.array([0.6, -0.3, 0.7, -0.5, 0.4, -0.6, 0.3, -0.6])
wage_index = 2 + 1.5 * experience + noise
print("Sample size:", len(experience))
print("Synthetic wage index:", wage_index)
```

## Why this matters

Before running an econometric package, it helps to understand what a regression coefficient represents. Ordinary least squares (OLS) chooses coefficients to minimize the sum of squared residuals. The eight observations here are *fictional teaching data*; they establish no real causal effect of experience on wages.

## Estimate intercept and slope

For a model with one explanatory variable, the slope is the sum of centered cross-products divided by the sum of squared centered explanatory values. The intercept makes the fitted line pass through the sample means.

```python
x_mean = experience.mean()
y_mean = wage_index.mean()
cross_product = np.sum((experience - x_mean) * (wage_index - y_mean))
x_sum_squares = np.sum((experience - x_mean) ** 2)
slope = cross_product / x_sum_squares
intercept = y_mean - slope * x_mean
print(f"Intercept: {intercept:.3f}; slope: {slope:.3f}")
```

## Residuals and goodness of fit

A fitted value is the model prediction, and a residual is observed minus fitted. With an intercept, OLS residuals sum to approximately zero. R-squared measures fit *within this sample*; it is not a measure of causality.

```python
fitted = intercept + slope * experience
residuals = wage_index - fitted
sse = np.sum(residuals**2)
sst = np.sum((wage_index - y_mean) ** 2)
r_squared = 1 - sse / sst
print("Residual sum:", round(residuals.sum(), 10))
print("R-squared:", round(r_squared, 4))
```

## Use the matrix form

Write `y = X beta + u` with an intercept column of ones. NumPy's least-squares solver is usually numerically preferable to computing an inverse directly. A model with collinear columns may not have uniquely identified coefficients.

```python
design = np.column_stack((np.ones(len(experience)), experience))
matrix_coefficients = np.linalg.lstsq(design, wage_index, rcond=None)[0]
assert np.allclose(matrix_coefficients, [intercept, slope])
print("Matrix solution agrees with the manual calculation.")
```

## Assumptions and interpretation

OLS is an algorithm for *sample association*. A causal interpretation additionally requires a defensible research design and identification assumptions, especially that the explanatory variable is not correlated with omitted determinants of the outcome. A high R-squared alone does not provide those assumptions.

For this one-regressor example, the slope has units of **wage-index points per experience year**, not a percentage change.

## Common errors

- A missing intercept changes the OLS objective and the interpretation of coefficients.
- Writing `x @ y` without checking array shapes can produce the wrong operation.
- A singular design matrix cannot uniquely determine all model parameters.
- An association in a fictional dataset is not an estimated treatment effect.

## Recap

- Simple OLS fits a line by minimizing squared residuals.
- The slope is a ratio of centered cross-products; the matrix form generalizes to more regressors.
- Fitted values and R-squared describe sample fit, not identification.
