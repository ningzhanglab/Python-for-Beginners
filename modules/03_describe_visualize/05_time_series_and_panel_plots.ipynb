## Learning goals

- **C**ompare trends across countries over time.
- **S**ummarize a binned scatter plot without hiding its construction.
- **L**abel a coefficient-style chart with units and uncertainty.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

panel = pd.DataFrame({
    "country": ["A", "A", "A", "B", "B", "B"],
    "year": [2022, 2023, 2024, 2022, 2023, 2024],
    "income_index": [100, 105, 111, 100, 102, 104],
    "life_index": [70, 71, 73, 65, 66, 67],
})
print("Fictional country-year rows:", len(panel))
```

## Why this matters

Country-year datasets may show distinct paths over time. Plotting each country's series helps catch unusual changes and compare directions. The indicators here are invented, not World Bank or other official statistics.

## Plot multiple country trends

Sort within each country before drawing lines so points are connected in chronological order. Marker shapes and a legend allow interpretation beyond color alone.

```python
fig, ax = plt.subplots(figsize=(6, 4))
for marker, (country, group) in zip(["o", "s"], panel.groupby("country")):
    ordered = group.sort_values("year")
    ax.plot(ordered["year"], ordered["income_index"],
            marker=marker, label=f"Country {country}")
ax.set(xlabel="Year", ylabel="Fictional income index (base 100)")
ax.legend(title="Country")
fig.tight_layout()
display(fig)
plt.close(fig)
```

## Calculate a transparent binned scatter

A binned scatter summarizes points within groups; it is not a regression estimator. For small teaching data, explicitly assign bins and compute means.

```python
binned = panel.copy()
binned["income_bin"] = pd.cut(binned["income_index"], bins=2)
bin_means = binned.groupby("income_bin", observed=True).agg(
    mean_income=("income_index", "mean"),
    mean_life=("life_index", "mean"),
)
print(bin_means.to_string())
```

## Plot the binned averages

Keep units and explain that bins are constructed from the fictional income index.

```python
fig_bin, ax_bin = plt.subplots(figsize=(6, 4))
ax_bin.scatter(bin_means["mean_income"], bin_means["mean_life"],
               color="#0072B2", marker="o")
ax_bin.set(xlabel="Fictional income index", ylabel="Fictional life index")
ax_bin.set_title("Binned averages, illustrative only")
fig_bin.tight_layout()
display(fig_bin)
plt.close(fig_bin)
```

## If you know Stata/R

| Task | Python | Stata | R |
| --- | --- | --- | --- |
| Time series | `ax.plot(year,value)` | `twoway line value year` | `geom_line(aes(year,value))` |
| Group means | `groupby(...).mean()` | `collapse (mean), by(group)` | `group_by(group) |> summarise(...)` |
| Bin labels | `pd.cut(series, bins=...)` | `egen cut()` | `cut()` |

Sorting the time variable first avoids misleading lines. Binned scatter plots are descriptive.

## Common errors

- Plotting unsorted years joins points in the wrong order.
- A panel may have missing years; verify balance before comparing growth rates.
- Using an observational binned scatter as a causal claim is inappropriate.
- If `pd.cut` yields missing bins, inspect bounds and missing values first.

## Recap

- Group and sort panel records before plotting trends.
- Use descriptive labels, markers, and units.
- Build binned plots transparently rather than treating them as automatic causal results.
