## Learning goals

- **C**reate labeled histogram, scatter, and line charts.
- **A**pply readable units and accessible visual encodings.
- **E**xport charts at publication-friendly resolution.

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

wages = pd.DataFrame({
    "region": ["North", "North", "South", "South", "East", "East"],
    "annual_wage": [42_000, 51_000, 36_000, 44_000, 62_000, 48_000],
    "experience_years": [2, 6, 4, 8, 10, 5],
})
print(wages.head(2))
```

## Why this matters

A chart can reveal an outlier, compare regional groups, or summarize change over time. Scientific figures should identify the measure, units, and source. These charts use fictional values; labels and captions should never imply they are real economic estimates.

## Start with figure and axes

A `Figure` is the canvas and an `Axes` is the plotting area. The object-oriented API makes labels and exporting explicit.

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(wages["annual_wage"], bins=4, edgecolor="black")
ax.set(xlabel="Annual wage (fictional USD)", ylabel="People (count)",
       title="Distribution of fictional wages")
fig.tight_layout()
plt.close(fig)
print("Histogram constructed")
```

## Compare two numeric variables

Use a scatter plot to see pairs of observations. Small samples should not be interpreted as a reliable economic relationship.

```python
fig_scatter, ax_scatter = plt.subplots(figsize=(6, 4))
ax_scatter.scatter(wages["experience_years"], wages["annual_wage"],
                   marker="o", color="tab:blue")
ax_scatter.set(xlabel="Experience (years)", ylabel="Annual wage (fictional USD)",
               title="Experience and wages, illustration")
fig_scatter.tight_layout()
plt.close(fig_scatter)
print("Scatter plot constructed")
```

## Plot a simple time series

Line plots are for values along an ordered axis. The example is an invented index, not a macroeconomic observation.

```python
years = np.array([2021, 2022, 2023, 2024])
index = np.array([100, 105, 108, 113])
fig_line, ax_line = plt.subplots(figsize=(6, 4))
ax_line.plot(years, index, marker="o", color="tab:blue")
ax_line.set(xlabel="Year", ylabel="Fictional price index (base 100)",
            title="Illustrative index over time")
fig_line.tight_layout()
plt.close(fig_line)
print("Line plot constructed")
```

## Export to PNG without committing outputs

Use a temporary directory for the exercise. In a research project, export to the documented `output/figures` path and regenerate from scripts.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    png_path = Path(folder) / "wage_histogram.png"
    fig.savefig(png_path, dpi=300, bbox_inches="tight")
    print("PNG created:", png_path.is_file())
```

## Common errors

- `NameError: plt is not defined` — import `matplotlib.pyplot as plt` first.
- Missing x/y labels make figures difficult to interpret; include units and data context.
- Figures overlap with notebook output if you forget `plt.close(fig)` when creating many figures.
- Avoid misleading line plots across unordered categories, and do not imply fictional data are official measurements.

## Recap

- Use Figure/Axes objects for reusable charts.
- Set clear titles and axes labels with units.
- Use accessible markers and labels rather than color alone.
- Save high-resolution figures from code to make results reproducible.
