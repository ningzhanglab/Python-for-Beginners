## Learning goals

- **Compare** distributions across categories using seaborn.
- **Add** clear labels and a caption to a statistical chart.
- **Visualize** an association without making a causal claim.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

survey = pd.DataFrame({
    "region": ["North", "North", "South", "South", "East", "East"],
    "wage": [42000, 51000, 36000, 44000, 62000, 48000],
    "experience": [2, 6, 4, 8, 10, 5],
    "employed": [True, True, True, True, True, True],
})
print("Fictional survey rows:", len(survey))
```

## Why this matters

A researcher needs to compare patterns among groups before choosing a model. Seaborn builds on matplotlib and helps create readable statistical graphics. The observations below are fictional, and any association in them is not evidence about actual workers.

## Plot distributions

A histogram shows the distribution of numeric data. Choosing the number of bins can change the visual impression, especially with six observations.

```python
fig, ax = plt.subplots(figsize=(6, 4))
sns.histplot(data=survey, x="wage", bins=4, ax=ax,
             color="tab:blue", edgecolor="black")
ax.set(xlabel="Annual wage (fictional USD)", ylabel="People (count)")
ax.set_title("Illustrative wage distribution")
fig.tight_layout()
display(fig)
plt.close(fig)
print("Histogram prepared")
```

## Compare group distributions

A boxplot shows quartiles and the median. With very small groups it is primarily a teaching device; it should not be used to infer population differences.

```python
fig_box, ax_box = plt.subplots(figsize=(6, 4))
sns.boxplot(data=survey, x="region", y="wage", ax=ax_box,
            color="lightsteelblue")
ax_box.set(xlabel="Fictional region", ylabel="Annual wage (fictional USD)")
ax_box.set_title("Wages by region (illustration)")
fig_box.tight_layout()
display(fig_box)
plt.close(fig_box)
print("Boxplot prepared")
```

## Draw a regression-style trend line

A fitted line summarizes the sample association, not a causal effect. Use distinct markers and explanatory labels so the chart remains understandable without color.

```python
fig_reg, ax_reg = plt.subplots(figsize=(6, 4))
sns.regplot(data=survey, x="experience", y="wage",
            ax=ax_reg, ci=None, scatter_kws={"marker": "o"})
ax_reg.set(xlabel="Experience (years)", ylabel="Wage (fictional USD)")
ax_reg.set_title("Fictional sample relationship")
fig_reg.tight_layout()
display(fig_reg)
plt.close(fig_reg)
print("Scatter + fitted line prepared")
```

## Common errors

- `ValueError: Could not interpret value` — check the exact DataFrame column spelling.
- A chart that lacks units leaves readers uncertain about the measure; label the axes.
- A fitted line does not establish causality; do not describe it as a treatment effect.
- A six-row fictional sample is not large enough for credible population-level conclusions.

## Recap

- `histplot` visualizes numeric distributions.
- `boxplot` compares numeric variables across groups.
- `regplot` displays a sample association with a fitted line.
- A descriptive caption and units are essential for reproducible communication.
