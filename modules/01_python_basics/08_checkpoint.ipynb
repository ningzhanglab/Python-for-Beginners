## Common errors

- Using nominal wages instead of adjusted wages changes the question.
- Confusing a decimal price index with a base-100 index can produce a 100-fold scaling mistake.
- Overwriting `sample` in a later cell produces hidden state; restart and run all cells.
- A small, synthetic sample cannot establish statistically meaningful population conclusions.
