## Learning goals

- **I**mport a standard-library module and call its functions.
- **I**nspect the availability of a research package.
- **D**istinguish an import from installing a dependency.

```python
import importlib.util
import math
import statistics

print('Installed NumPy:', importlib.util.find_spec('numpy') is not None)
print('Sample square root:', math.sqrt(4))
```

## Why this matters

Economists reuse tested functionality instead of rewriting every calculation. A **module** is an importable Python file or collection of functions. A **package** organizes modules and may require installation in a project environment.

## The standard library

Python includes common modules without extra installation. `statistics.mean` is useful for small sequences, while `math` offers numerical functions for scalar values.

```python
wages = [42000, 48000, 51000, 48000]
mean_wage = statistics.mean(wages)
print('Mean wage:', mean_wage)
print('Square root of 10000:', math.sqrt(10000))
```

## Package imports and aliases

NumPy is a third-party package pinned by the course environment. Use the common alias `np` for its API.

```python
import numpy as np

wage_array = np.array(wages)
print('NumPy mean:', wage_array.mean())
print('NumPy version:', np.__version__)
```

## Finding documentation

A function's docstring and official reference explain its inputs, outputs, and limitations. Use `help` or `.__doc__` when offline.

```python
print('Mean documentation starts:')
print(statistics.mean.__doc__.splitlines()[0])
print('Pandas import availability:', importlib.util.find_spec('pandas') is not None)
```

## Common errors

- `ModuleNotFoundError`: the package is not installed in your selected Python environment. Use `uv sync --locked`.
- `AttributeError`: the module may not have a function with that spelling; consult its documentation.
- A Jupyter kernel uses a different environment from your terminal: select the project's `.venv` kernel.

## Recap

- Modules expose reusable functions; packages organize code and dependencies.
- Installing packages and importing them are different operations.
- Use module documentation rather than guessing function signatures.
