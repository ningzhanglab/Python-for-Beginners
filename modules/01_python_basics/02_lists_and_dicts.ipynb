## Why this matters

Research datasets contain groups of related observations. Before using pandas, you need a few simple Python **containers**: a list for an ordered collection, a dictionary for named lookups, a tuple for a fixed group, and a set for unique values.

## Lists, indexing, and slicing

Python counts positions from zero. A list can change, so you can append observations. Slices select a range of positions and exclude the endpoint.

```python
wages = [42000, 48000, 51000, 48000]
print('First wage:', wages[0])
print('First two wages:', wages[:2])
print('Last wage:', wages[-1])
print('Average:', sum(wages) / len(wages))
```

## Dictionary lookups

A dictionary maps **keys** to values. Keys make country indicators easier to interpret than positional list entries.

```python
country_growth = {'Kenya': 5.1, 'Ghana': 3.8, 'Rwanda': 6.2}
print('Kenya growth:', country_growth['Kenya'])
print('Available countries:', list(country_growth))
```

## Tuples, sets, and mutability

A tuple is an ordered collection you generally leave unchanged. A set keeps only distinct entries, without a meaningful order. A list is mutable: editing it can change later results.

```python
survey_period = (2023, 2024)
regions = ['North', 'South', 'North', 'West']
unique_regions = set(regions)
print('Period:', survey_period)
print('Unique regions:', sorted(unique_regions))
print('Unique wages:', sorted(set(wages)))
```

## If you know Stata/R

| Task | Python | Stata | R |
| --- | --- | --- | --- |
| Keep first two positions | `wages[:2]` | `wage in 1/2` (observation selection) | `wages[1:2]` |
| Named lookup | `d['Kenya']` | scalar/macro or dataset lookup | `d[['Kenya']]` |
| Unique entries | `set(regions)` | `levelsof region` | `unique(regions)` |

Unlike R, Python lists are indexed from zero. Stata data columns are different from Python lists.

## Common errors

- `IndexError: list index out of range`: you requested a position that does not exist; remember that indexing begins at zero.
- `KeyError: 'Country'`: the dictionary key does not match exactly; inspect `dict.keys()`.
- Editing a list unexpectedly changes it elsewhere: lists are mutable; copy with `list(original)` when needed.

## Recap

- Lists are ordered and mutable; tuples are ordered and typically unchanged.
- Dictionaries map keys to values, useful for indicator lookups.
- Sets remove duplicates, and slices select ranges without including their endpoint.
