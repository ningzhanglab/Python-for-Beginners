## Why values and variables matter

A wage survey may store pay as a number, region as text, and an employment indicator as true or false. Before cleaning a dataset, it helps to know what each **value** represents and which **type** Python gives it.

## Numbers, text, Boolean values, and missingness

An `int` is an integer, a `float` holds a decimal number, a `str` holds text, a `bool` holds `True` or `False`, and `None` marks the absence of a value. `None` is not zero.

```python
annual_wage = 48000
price_index = 120.0
region = "North"
is_employed = True
missing_response = None
for value in (annual_wage, price_index, region, is_employed, missing_response):
    print(repr(value), type(value).__name__)
```

## Arithmetic for empirical work

An annual growth rate compares a new value to an old one. For a change from 50,000 to 52,000, use the ratio minus one.

```python
wage_2023 = 50000
wage_2024 = 52000
growth_pct = (wage_2024 / wage_2023 - 1) * 100
real_wage = wage_2024 / price_index * 100
print(f"Nominal wage growth: {growth_pct:.1f}%")
print(f"Real wage in base-index units: {real_wage:,.2f}")
```

## Why floating-point arithmetic surprises people

Binary computers cannot represent every decimal fraction exactly. `0.1 + 0.2` is not represented as precisely `0.3`. Use a tolerance for numeric checks rather than exact float equality.

```python
print(0.1 + 0.2)
print(math.isclose(0.1 + 0.2, 0.3, rel_tol=1e-12))
```

## Text cleaning

Survey labels can differ because of spaces and capitalization. `.strip()` removes extra whitespace; `.lower()` changes case; `.replace()` substitutes text; `.split()` separates words.

```python
region_raw = "  NORTH "
region_clean = region_raw.strip().lower()
question = "income-per-year"
print(region_clean)
print(question.replace("-", " ").split())
print(f"Recorded region: {region_clean}")
```

## Common errors

- `NameError: name 'annual_wage' is not defined` — rerun the cell assigning that name.
- `TypeError: can only concatenate str (not "int") to str` — combine text with an f-string or convert the number first.
- `ValueError: could not convert string to float` — check that your survey entry is numeric before converting.
- Unexpected decimal rounding — floating-point representation is approximate; use `math.isclose` for checks.

## Recap

- Growth rates compare final and initial amounts; real values adjust for a price index.
