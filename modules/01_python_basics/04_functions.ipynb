## Why this matters

Research projects repeat the same calculation across countries, years, and papers. A function gives that operation a name, defines its inputs, and returns its result. A **docstring** records what the function expects and returns.

## Define a deflator

A price index of 125 with a base of 100 converts a nominal amount into base-year purchasing-power units. This is a simplified index adjustment, not a survey design or cost-of-living model.

```python
def deflate(nominal, price_index, base=100):
    """Return a nominal amount in price-index base-year units."""
    if price_index <= 0:
        raise ValueError('price_index must be positive')
    return nominal / price_index * base

print('Deflated amount:', deflate(63000, 126))
```

## Compound annual growth rate

CAGR summarizes multiplicative growth across years. It only makes sense here with positive endpoints and a positive number of years.

```python
def cagr(start, end, years):
    """Return compound annual growth as a decimal, not percent."""
    if start <= 0 or end <= 0 or years <= 0:
        raise ValueError('Inputs must be positive')
    return (end / start) ** (1 / years) - 1

print('Annualized growth:', cagr(100, 121, 2))
```

## Check functions with known examples

Tests make code safer to reuse. `math.isclose` compares floating-point results within a small tolerance.

```python
assert math.isclose(deflate(63000, 126), 50000)
assert math.isclose(cagr(100, 121, 2), 0.10)
print('Both numerical checks passed.')
```

## Common errors

- `TypeError: missing required positional argument`: pass all required inputs when calling a function.
- A function prints a result but returns `None`: use `return` for reusable calculations.
- `ZeroDivisionError`: dividing by a zero price index or population is invalid; validate inputs.
- Using an annual growth number as a percentage when it is a decimal can change interpretation by a factor of 100.

## Recap

- Use `def`, arguments, and `return` to package calculations.
- Defaults such as `base=100` can make a function convenient but should be documented.
- Check a small known example before using a function on real data.
