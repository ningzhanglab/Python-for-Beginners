## Why it matters

You may have a table of countries, a questionnaire or a spreadsheet of wages. Python lets you record the steps between that table and a result. **Reproducible** means another person can follow those same steps with the same inputs and obtain the same result. It does not mean the research question or the data must be correct. This course teaches you to inspect both.

Read this page first, then use the notebook lesson to get comfortable with Colab. The first hour needs no local installation. Small examples use made-up numbers for fictional countries; a separate historical country panel will keep returning as your skills grow.

## Worked example

### Start with a question, then a calculation

Our running question is **Do people live longer in richer countries?** It leads to measurable quantities: life expectancy at birth (years) and GDP per person (average economic production per person). Before drawing a chart, practise a subtraction. In fictional Northland, life expectancy changes from 68 to 74 years. In Colab, click the run triangle beside this code cell; on the website, read its tested output.

```python
print(74 - 68)
```

The output is 6: an increase of six **years**, not six percent. The computer follows your instructions exactly; it cannot infer units from the numbers. A **code cell** is one block of instructions. **Output** is the result shown beneath it.

### Know what a number refers to

A measurement needs a place, a date and a definition. The next prepared cell prints one real observation from our historical country panel. `print` is an instruction to display what is inside its parentheses. Quotation marks enclose text; the decimal number is a measurement.

```python
print("Canada, 2007: life expectancy at birth (years)")
print(80.653)
```

Notice the country, year and units together. These values are from the Gapminder teaching extract, not today's statistics. It covers 142 countries every five years from 1952 to 2007. A country average does not describe each person's experience. Later lessons load the source instead of typing individual values.

### See a simple comparison

Fictional Eastria has GDP per person of 30,000 dollars and fictional Northland has 15,000. Divide to compare their levels. These are made-up numbers.

```python
print(30000 / 15000)
```

The ratio is 2.0: Eastria's level is twice Northland's. That comparison does not establish why the countries differ in income or life expectancy. It is a descriptive calculation, meaning a description of the observed numbers.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### Numbers and text are different

Quotation marks turn digits into text. This deliberately attempts subtraction using text where a number is needed.

```python
try:
    print(74 - "68")
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

A TypeError says the kinds of inputs do not fit the operation. The numeric calculation uses 68 without quotation marks. Lesson 1.1 develops this distinction.
