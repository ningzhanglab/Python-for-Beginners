## Why it matters

Learning Python is partly learning how to recover when something surprises you. Keep one research question in view, make one change at a time, and check the output against your expectation. You do not need to memorize every command. You need to know what an instruction is meant to do and how to test that it did it.

Use this routine for each lesson: read the learning goals; predict a worked example's result; run it; attempt the exercises in new cells; then compare with the solutions. Pause after 20–40 minutes. A short return visit often helps more than repeatedly running a confusing cell.

## Worked example

### Predict, run, explain

These are made-up numbers for Northland. Before running, predict the change from 69 to 73 years. Assigning with `=` gives a value a name; `print` displays a value.

```python
earlier = 69.0
later = 73.0
gain = later - earlier
print(gain)
```

You should see 4.0 years. Say the result in a sentence: “Life expectancy rose by four years in this invented scenario.” If the output differs from your prediction, look at the inputs before changing the formula.

### Editing is not rerunning

A notebook runtime remembers values from cells you ran. Editing the earlier cell does not rerun later calculations automatically. This cell changes `later` but initially prints the previously stored `gain`.

```python
later = 74.0
print(gain)
gain = later - earlier
print(gain)
```

The first output is still 4.0; only after recalculation does it become 5.0. Use **Runtime → Restart session and run all** in Colab (wording may vary) when checking a finished notebook. Save your copy in Drive first if you want to keep edits. A saved notebook and a live runtime are different things.

### Make a claim checkable

Here is a small, real calculation from the running example. Canada's life expectancy was 79.770 years in 2002 and 80.653 in 2007 in the historical extract. Store the change and round only the displayed result. `round(value, 3)` shows three decimal places of precision in this case.

```python
canada_gain = 80.653 - 79.770
print(round(canada_gain, 3))
```

The increase is 0.883 years over five years. The input values, date range and subtraction together let someone check your statement. Rounding a display is different from changing your source data. Exact source values are in [the data documentation](https://github.com/ningzhanglab/Python-for-Beginners/blob/main/data/README.md).

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### A cell uses a name you have not defined

This misspelling is deliberate. Look at the final line of the displayed error.

```python
try:
    print(canada_gaim)
except NameError as error:
    print(f"{type(error).__name__}: {error}")
```

The message identifies `canada_gaim`; the defined name is `canada_gain`. Rerun the defining cell after correcting the name. If many names are missing after a disconnect, restart at the top.

### A pasted number contains text

A copied spreadsheet value may look numeric while still being text. This demonstration uses quoted text in a subtraction.

```python
try:
    print("80.653" - 79.770)
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

Keep the number numeric in this small example. Later, lesson 2.4 teaches careful conversion of a whole column, including values that cannot be converted.
