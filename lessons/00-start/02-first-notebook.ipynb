## Worked example

### 1. Open a notebook

Choose **Open in Colab** above. A **cell** is one block of text or code. Code cells have a run button on the left; text cells explain what you are doing. If you want to edit a saved copy, choose **File → Save a copy in Drive**. Your changes belong to your copy.

### 2. Ask Python to calculate

Subtract the two fictional life expectancies. The last calculation in a cell is displayed below it.

```python
73.5 - 68.0
```

Notice `5.5`: an increase of 5.5 years. Output means the result produced by a cell. The inputs above are made-up numbers, not estimates for a real country.

### 3. Give a value a name

A **variable** is a name referring to a value. The equals sign stores the value on its right under the name on its left. Run this cell before the cells that use these names.

```python
life_exp_2000 = 68.0
life_exp_2020 = 73.5
change = life_exp_2020 - life_exp_2000
print(change)
```

Again you should see `5.5`. A **function** is a named operation: `print(...)` displays whatever you put between its parentheses. The names make the meaning of the subtraction visible.

A line beginning with `#` is a **comment**, a note Python ignores. Change the end-year value to explore a different made-up scenario. Here we use 74.0.

```python
# Northland: a different made-up scenario
life_exp_2020 = 74.0
change = life_exp_2020 - life_exp_2000
print(change)
```

The change is now `6.0`. A notebook remembers values from cells you already ran. Editing a cell does not change the remembered value until you run it again. For a reliable check, use **Runtime → Restart session and run all** (the wording may vary slightly) and read the outputs from the top.

### 4. See where you are heading

Our running question is **“Do people live longer in richer countries?”** Here “richer” means higher GDP per person: the value of economic production divided by population. It is a country average, not a measure of each person's income.

Run the prepared setup cell below. It puts the small teaching dataset in this notebook's working folder. In the course build it uses the committed CSV; in a fresh Colab session it downloads a fixed public source and checks its contents. You do not need to understand each line yet. It contains no paid service, Google Drive connection or private data.

```python
# Prepared setup: use the course CSV locally; fetch its pinned source in Colab.
from pathlib import Path
import hashlib
import io
from urllib.request import urlopen
import pandas as pd

data_file = Path("data/gapminder.csv")
if not data_file.exists():
    source = "https://raw.githubusercontent.com/jennybc/gapminder/391b5a40d3574c3b89d889dd7d72f7f3fdb4a70c/inst/extdata/gapminder.tsv"
    with urlopen(source, timeout=30) as response:
        raw = response.read()
    expected = "f74275b4a2f4a7c8e3a09a752f9d6e43d1759d78d5cbfdebaaa3e1a6efc68f78"
    if hashlib.sha256(raw).hexdigest() != expected:
        raise ValueError("The data download changed. Please report this to the course maintainer.")
    downloaded = pd.read_csv(io.BytesIO(raw), sep="\t", dtype=str)
    downloaded = downloaded.rename(columns={"continent": "region", "lifeExp": "life_exp", "pop": "population", "gdpPercap": "gdp_pc"})
    data_file.parent.mkdir(parents=True, exist_ok=True)
    downloaded.to_csv(data_file, index=False, lineterminator="\n")
print("Course data ready: data/gapminder.csv")
```

Look for **Course data ready**. An internet connection is needed for the first download. Colab's working files are temporary; after a runtime reset, rerun this setup cell.

Run the next prepared cell to draw a **scatter plot**, a chart with one point for each country in 2007. The two named libraries (collections of reusable functions), pandas and matplotlib, load the table and draw the chart. Later lessons unpack these commands.

```python
#| label: fig-income-life
#| fig-alt: "Scatter plot of 142 countries in 2007. Life expectancy generally rises with GDP per person, with substantial differences between countries at similar income levels."
#| fig-cap: "Historical teaching data: Gapminder via Jennifer Bryan (CC0), 2007. Each point is one country; countries are not weighted by population."
import matplotlib.pyplot as plt

countries = pd.read_csv(data_file)
year_2007 = countries[countries["year"] == 2007]
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.scatter(year_2007["gdp_pc"], year_2007["life_exp"],
           color="#21645a", edgecolor="white", linewidth=0.5, alpha=0.8)
ax.set_xscale("log")
ax.set_xlabel("GDP per person (inflation-adjusted US dollars; log scale)")
ax.set_ylabel("Life expectancy at birth (years)")
ax.set_title("Do people live longer in richer countries? · 2007")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

Notice the broad upward pattern and the spread around it. The horizontal axis uses a **log scale**: equal steps represent equal ratios (such as doubling), rather than equal dollar increases. No colour key is needed; all countries use the same marker.

This picture shows an **association**, a pattern in which two measurements vary together. It does not show that making a country richer causes people to live longer. Public health, education and many other differences could matter. These are 142 countries in an historical extract, not all countries or today's values. ## When it goes wrong

### “NameError: name … is not defined”

A **NameError** means Python does not know the name you used. Perhaps it is misspelled, or the cell that defines it has not run. The following demonstration deliberately uses a misspelling. The `try`/`except` wrapper catches the expected error so the rest of the lesson can run; you will learn that wrapper later.

```python
try:
    print(life_expec_2020)
except NameError as error:
    print(f"{type(error).__name__}: {error}")
```

The message names `life_expec_2020`, which differs from `life_exp_2020`. Read the last line of a longer error report first, then find the line it points to. Fix the spelling and rerun the defining cell if necessary.

```python
print(life_exp_2020)
```

In an unedited run you should see 74.0. If you changed the input earlier, your output may differ.

### “TypeError” when mixing text and numbers

A **type** is the kind of value, such as number or text. Quotation marks create text. This deliberately tries to subtract text from a number and prints the expected error.

```python
try:
    print(73.5 - "68.0")
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

The values need compatible types for subtraction. Use the numeric value `68.0` without quotation marks, as in the first example.

If Colab disconnects or reports that `countries` is undefined, reconnect and run the notebook from the beginning. Saving a notebook saves its cells; it does not keep a runtime alive forever.
