## Why it matters

A local environment gives one project its own Python packages and persistent files. An **interpreter** runs Python code; a notebook **kernel** is the live process running your cells. Installing a package into one interpreter does not make it available in every kernel. You can complete this lesson in Colab without installing anything. The made-up numbers below are for a small environment check, then we inspect the real country data.

## Worked example

### Inspect the session you already have

A terminal is a window for commands to the operating system. This code cell is Python, not a terminal command. Print the interpreter path and Python version before installing or debugging anything. The path is local to your session and will differ across computers; remove private folder names from a public bug report.

````python
import sys
from pathlib import Path
print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Working folder:", Path.cwd().name)
````

Python 3.12 is the course target. A file named python can belong to several environments; the complete sys.executable path identifies the one actually running. A working folder is the starting point for relative file paths. In Colab it is a temporary server folder, not your laptop’s Downloads folder.

### Get one project folder and the shared requirements

Open a terminal **in that extracted folder**. In VS Code, **File → Open Folder**, then **Terminal → New Terminal** does this without typing an absolute path. Both systems use:

```sh

```

The first command creates the project's `.venv` folder, downloading a compatible Python 3.12 patch version if needed. The second installs exactly the package list from the course file. On macOS/Linux, verify it with:

```sh
.venv/bin/python --version
uv pip check
```

On Windows PowerShell:

```powershell
.venv\Scripts\python --version
uv pip check
```

`uv pip check` reports incompatible installed package requirements; it does not evaluate your analysis. Using the environment's Python explicitly avoids relying on whichever `python` command your terminal happens to find.

### Check a package and a tiny calculation

Import pandas in the current kernel and summarize made-up Northland measurements. The result checks that this kernel can use the package; it does not prove every terminal uses the same Python.

````python
import importlib.metadata
import pandas as pd
print("pandas:", importlib.metadata.version("pandas"))
practice = pd.Series([68.0, 71.0, 74.0], name="Northland life expectancy")
practice_mean = practice.mean()
print("Made-up mean, years:", practice_mean)
````

The mean is 71 years. An installed package version and a plausible calculation are useful checks together. If you see ModuleNotFoundError, first confirm the selected kernel and the environment where installation ran.

### Prepare the historical country panel

This prepared cell uses the local course CSV or downloads its checksum-verified public source in a fresh browser session. A checksum identifies exact file contents; changed bytes stop the preparation.

````python
# Prepared setup: use the course CSV locally; fetch its pinned source in Colab.
from pathlib import Path
import hashlib
import io
from urllib.request import urlopen
import pandas as pd

data_file = Path("data/gapminder.csv")
if not data_file.exists():
    source = "https://raw.githubusercontent.com/jennybc/gapminder/391b5a40d3574c3b89d889dd7d72f7f3fdb4a70c/inst/extdata/gapminder.tsv"
    with urlopen(source, timeout=30) as response:
        raw = response.read()
    expected = "f74275b4a2f4a7c8e3a09a752f9d6e43d1759d78d5cbfdebaaa3e1a6efc68f78"
    if hashlib.sha256(raw).hexdigest() != expected:
        raise ValueError("The data download changed. Please report this to the course maintainer.")
    downloaded = pd.read_csv(io.BytesIO(raw), sep="\t", dtype=str)
    downloaded = downloaded.rename(columns={"continent": "region", "lifeExp": "life_exp", "pop": "population", "gdpPercap": "gdp_pc"})
    data_file.parent.mkdir(parents=True, exist_ok=True)
    downloaded.to_csv(data_file, index=False, lineterminator="\n")
print("Course data ready: data/gapminder.csv")
````

Look for **Course data ready**. This is a historical teaching extract: 142 countries, twelve five-year observations from 1952 to 2007. Its GDP measure is inflation-adjusted source dollars with no documented base year. It is not current statistics.

### Check a real input after choosing the kernel

Read the country panel relative to the current working folder, filter 2007 and check its size. This is a data-location check as well as a package check.

````python
countries = pd.read_csv(data_file)
local_sample = countries.loc[countries["year"] == 2007]
print("2007 country observations:", len(local_sample))
print("Columns:", countries.columns.tolist())
````

Expect 142 countries and the six documented source columns. If that fails, fixing a path is different from reinstalling pandas. The prepared setup makes this cell work in the browser too; no installation step is run by this notebook.
