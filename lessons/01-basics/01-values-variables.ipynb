## Why it matters

A notebook full of unexplained numbers is hard to check. Variables let you name the quantities in a research question, such as a country's population or a respondent's employment status. A **variable** is a name referring to a value; a **type** describes the kind of value. We begin with made-up numbers for fictional Northland before storing a real country observation.

## Worked example

### Numbers and units

Use an equals sign to assign a value. `gdp_per_person` is a name you choose; underscores join words because Python names cannot contain spaces. `#` introduces a comment that Python ignores. An underscore inside a number, such as `5_500_000`, is just a readability aid.

```python
gdp_per_person = 42000
population = 5_500_000
# Made-up Northland values: dollars per person multiplied by people
gdp = gdp_per_person * population
print(gdp)
```

The result is 231000000000 dollars of production in this invented scenario. Multiplying dollars per person by people leaves dollars. Python does not track the units automatically.

### Text and True/False

A **string** is text enclosed in quotation marks. A **Boolean** is a True/False value, useful for yes/no questions. `type(...)` displays a value's type. `int` means whole number, `float` means a number with a decimal representation, `str` means text and `bool` means Boolean.

```python
country_name = "Northland"
life_expectancy = 73.5
employed = True
print(type(population).__name__)
print(type(life_expectancy).__name__)
print(type(country_name).__name__)
print(type(employed).__name__)
```

Notice `int`, `float`, `str` and `bool`. The two underscores around `__name__` are part of this prepared way to display the type's label; you do not need to use that detail in ordinary calculations. `True` has a capital T; quoted `"True"` would be text.

### Assignment and comparison

`=` assigns; `==` asks whether values are equal. `>` asks whether the left value is greater. Both comparisons return a Boolean.

```python
print(population == 5_500_000)
print(life_expectancy > 70)
print(country_name == "northland")
```

The first two results are True and the last is False. Text comparisons distinguish capitals. A threshold such as 70 is your analytic choice, not a scientific standard supplied by Python.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### Unknown name

Variable names must match exactly, including capitals.

```python
try:
    print(Canada_life_exp)
except NameError as error:
    print(f"{type(error).__name__}: {error}")
```

The defined name starts with lowercase `canada`. Python treats `Canada_life_exp` as a different name. Descriptive, consistently lowercase names help prevent this.

### Arithmetic with quoted digits

This deliberately uses a numeric-looking string in subtraction.

```python
try:
    print("73.5" - 68.0)
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

Use numeric values for arithmetic. The string-to-number function `float("73.5")` can convert valid numeric text; inspect the source before converting real data.
