## Why it matters

Research data usually contain many values and several attributes for each observation. A **list** stores values in an order. A **dictionary** pairs names, called keys, with values. These small structures prepare you for table columns and records in Part 2. The first examples use made-up numbers for fictional Northland and an invented survey respondent.

## Worked example

### A list is an ordered collection

Square brackets create a list. Commas separate its values. Here the order corresponds to years 2000, 2010 and 2020; keep that meaning in the comment.

```python
# Northland life expectancy in 2000, 2010, 2020: made-up numbers
life_exp_values = [68.0, 71.0, 74.0]
print(life_exp_values)
print(len(life_exp_values))
```

There are three values. `len(...)` counts items. A list does not remember the dates by itself; ordering is part of your responsibility.

### Positions start at zero

An **index** here means a position in the list. Position 0 is the first item, position 1 the second. Retrieve the first and third values, then subtract.

```python
print(life_exp_values[0])
print(life_exp_values[2])
print(life_exp_values[2] - life_exp_values[0])
```

The values are 68.0 and 74.0; the increase is six years. Brackets after a list ask for an item, while brackets around several comma-separated values create a list.

### A dictionary names fields

Braces create a dictionary; a colon connects each key to its value. This fictional respondent has an age, years of education and a Boolean employment status. Retrieve education by its key.

```python
respondent = {"age": 34, "education_years": 16, "employed": True}
print(respondent["education_years"])
print(respondent["employed"])
```

You see 16 and True. Unlike a list position, a key can express the meaning of the measurement. One key identifies one value in a dictionary; repeated keys do not create repeated observations.

### Store the running example as a record

**The running example.** The real observations below come from the course's [historical Gapminder teaching extract](https://github.com/ningzhanglab/Python-for-Beginners/blob/main/data/README.md), not the fictional examples. Canada in 2007 has life expectancy of 80.653 years and GDP per person of 36,319.23501 inflation-adjusted US dollars in this extract. GDP per person measures average economic production, not each person's earnings. These are historical observations, not today's statistics or evidence of causation.

Keep the country, year and two measurements together, then print the life-expectancy field.

```python
canada = {"country": "Canada", "year": 2007, "gdp_pc": 36319.23501, "life_exp": 80.653}
print(canada["country"], canada["year"])
print(canada["life_exp"])
```

The named record is easier to interpret than a bare list of numbers. Part 2 combines many country-year records into a table.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### A position is outside the list

There are three items, so the valid positions are 0, 1 and 2.

```python
try:
    print(life_exp_values[3])
except IndexError as error:
    print(f"{type(error).__name__}: {error}")
```

IndexError means the requested position is unavailable. Compare the position with `len(life_exp_values)`. The count is not the final index.

### A dictionary key is absent

This deliberately uses a key that was not defined.

```python
try:
    print(canada["GDP"])
except KeyError as error:
    print(f"{type(error).__name__}: {error}")
```

KeyError identifies the missing key. Use the actual `"gdp_pc"` key. In Part 2 the same error often means a missing or misspelled table column.

## Recap

- Lists keep order; dictionaries keep named fields.
- List positions begin at zero.
- Dictionary keys must match exactly.
- Neither structure supplies units or source information automatically.
