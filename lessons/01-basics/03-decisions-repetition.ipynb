## You will learn

## Why it matters

A cleaning rule often says “if a value meets this condition, do this.” A repeated task often says “do the same step for each item.” Python expresses these ideas with `if` and `for`. We start with made-up numbers and a fictional threshold, then repeat a calculation on Canada's historical observations. A rule you choose for teaching is not an official classification.

## Worked example

### Choose between two messages

A **condition** is a test that evaluates to True or False. `if` runs its indented block when the test is True; `else` runs the other block. A block is a group of lines belonging together. The colon begins the block, and four leading spaces indent each line inside it.

```python
# Made-up Northland value; 70 is an illustrative threshold
life_exp = 73.5
if life_exp >= 70:
    print("At least 70 years")
else:
    print("Below 70 years")
```

The output selects “At least 70 years.” `>=` includes equality. Change the value to 70 and then to 69 to test the boundary; restore the cell before running the remaining worked examples.

### Repeat one operation

A **loop** repeats instructions. `for` takes each item in a list in order, gives it a temporary name, and runs the indented block. Here each invented life expectancy receives a gap from 80 years. The line after the loop is not indented, so it runs once.

```python
fictional_life_exp = [68.0, 73.5, 78.0]
for value in fictional_life_exp:
    print(80 - value)
print("Finished all three values")
```

The gaps are 12.0, 6.5 and 2.0 years, followed by one final message. The loop name `value` takes a different number on each repetition. It is a name, not a special Python keyword.

### Accumulate a total

Initialize a total before the loop. On each repetition, add the next value and store the updated total. Divide by the count to obtain a simple average.

```python
total = 0
for value in fictional_life_exp:
    total = total + value
fictional_mean = total / len(fictional_life_exp)
print(round(fictional_mean, 2))
```

The average is 73.17 years after rounding. Resetting `total` inside the loop would erase earlier additions. This is the arithmetic mean of three invented numbers, not a population-weighted result.

### Repeat on the running example

Canada's life expectancies in 2002 and 2007 are 79.770 and 80.653 in the historical extract. Repeat the threshold comparison for these two real observations. Both refer to Canada, not two different countries.

```python
canada_life_exp = [79.770, 80.653]
for measurement in canada_life_exp:
    print(measurement >= 80)
```

The results are False and True. Canada crosses the illustrative threshold between these observations. This describes a comparison; it does not explain what caused the change.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### A line is not indented

The next prepared example stores intentionally broken code as text. `exec` asks Python to read that text as instructions; the wrapper catches the resulting indentation error. This is only for demonstrating a mistake, not a recommended way to run ordinary code.

```python
try:
    exec('if True:\nprint("At least 70")')
except IndentationError as error:
    print(f"{type(error).__name__}: {error}")
```

The line after the colon must be indented. Use four spaces consistently; do not mix tabs and spaces. Compare with the correctly indented worked example.

### The colon is missing

This prepared example again reads a short broken command from text.

```python
try:
    exec('for measurement in [79.770, 80.653]\n    print(measurement)')
except SyntaxError as error:
    print(f"{type(error).__name__}: {error}")
```

SyntaxError means Python cannot read the command structure. A for or if header needs a colon at the end. Fix the header before changing the data.
