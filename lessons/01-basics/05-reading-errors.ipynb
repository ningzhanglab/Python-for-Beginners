## Why it matters

An **exception** is Python's way of saying an instruction could not be completed. A **traceback** is a report of where execution reached that problem. Start with its final line, which usually names the exception and describes it; then find the source line it points to. Fix one cause, rerun, and inspect the new result. An error is not evidence that you are unsuited to programming.

These demonstrations use made-up numbers for fictional countries. The examples catch their intended exceptions so you can run the whole lesson. In your ordinary analysis, unexpected exceptions should stop execution until you understand them.

## Worked example

### Read the bottom line first

This first example deliberately misspells a variable. `try` attempts the indented instruction. `except NameError` catches only that exception. `type(error).__name__` displays the exception's name; the text after it is the message.

```python
northland_life_exp = 74.0
try:
    print(northland_life_expp)
except NameError as error:
    print(f"{type(error).__name__}: {error}")
```

The message names `northland_life_expp`. Check spelling, capitals and whether the defining cell ran. A caught exception is not a repaired calculation; it just lets the demonstration continue.

### Repair the name and inspect the result

Use the exact defined name.

```python
print(northland_life_exp)
```

The result is 74.0. When troubleshooting a real notebook, changing a name is only a fix if the resulting value has the meaning and units you intended.

### Look at the kinds of inputs

A copied survey response can look like a number but be text. Inspect the type before converting valid numeric text with `float` (a decimal-number conversion function).

```python
age_text = "34"
print(type(age_text).__name__)
age_number = float(age_text)
print(age_number + 1)
```

The original type is str; the corrected calculation gives 35.0. Converting text is appropriate only if it actually represents a numeric measurement. Lesson 2.4 handles invalid and missing column values explicitly.

### Identify the exact source line

A **SyntaxError** means Python could not read the command structure. An **IndentationError** is one kind of syntax problem. `compile` checks code stored as text without running it. The next prepared snippet omits indentation; the exception provides a line number and the offending text.

```python
broken = "if True:\nprint(74.0)"
try:
    compile(broken, "practice", "exec")
except IndentationError as error:
    print(type(error).__name__)
    print("Line:", error.lineno)
    print("Text:", error.text.strip())
    print("Message:", error.msg)
```

The report identifies line 2, where the print line should be indented. The filename `practice` is just the label supplied for this example. In a long traceback from a library, look for the notebook line you wrote before editing library internals.

### Check a real record's keys

The running example stores Canada's historical life expectancy in a named record. `.keys()` lists the available fields. Inspect them, then retrieve the actual key.

```python
canada = {"country": "Canada", "year": 2007, "life_exp": 80.653}
print(list(canada.keys()))
print(canada["life_exp"])
```

The record has `country`, `year` and `life_exp`, not a key named `life_expectancy`. Part 2 uses the same diagnosis for missing table columns. The real measurement is historical and country-level.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### KeyError: the label is absent

Deliberately request an undefined field.

```python
try:
    print(canada["life_expectancy"])
except KeyError as error:
    print(f"{type(error).__name__}: {error}")
```

The missing label is printed. Check the real keys rather than inventing a replacement measurement. Here the corrected key is `life_exp`.

### FileNotFoundError: the location is absent

`Path` describes a location; `.read_text()` reads a text file. `TemporaryDirectory` supplies an empty disposable folder, so this demonstration cannot accidentally read a real private file. The file will deliberately be missing.

```python
try:
    from pathlib import Path
    from tempfile import TemporaryDirectory
    with TemporaryDirectory() as folder:
        (Path(folder) / "missing-survey.csv").read_text()
except FileNotFoundError as error:
    print(f"{type(error).__name__}: {error}")
```

Check the working folder, spelling, file ending and whether a Colab upload survived a reset. Do not share personal folder paths or confidential file contents when asking for help. Loading tables is introduced in lesson 2.1.

## Recap

- Read the exception type and message, then find the indicated source line.
- NameError concerns names; TypeError concerns incompatible kinds of values; KeyError concerns labels.
- Indentation and syntax errors concern code structure; FileNotFoundError concerns a location.
- A caught error is a demonstration, not a substitute for fixing an analysis.
