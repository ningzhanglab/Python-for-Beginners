## Why it matters

Repeating a formula by copying it can introduce inconsistent edits. A **function** is a named operation you can call with different inputs. The input values supplied in a call are **arguments**. Naming a calculation lets you check its definition once and then reuse it. We start with made-up numbers and calculate historical Canadian GDP growth only after checking the formula on a simple case.

## Worked example

### Use functions Python already provides

`sum` adds a list of numbers, `len` counts its items, and `round` rounds a displayed result. These made-up Northland survey incomes are dollars per month. Find their arithmetic mean.

```python
monthly_incomes = [1000, 1500, 2000]
mean_income = sum(monthly_incomes) / len(monthly_incomes)
print(round(mean_income, 2))
```

The average is 1500.0 dollars per month. Parentheses call a function. When a function needs more than one input, commas separate them: `round(mean_income, 2)` supplies the value and number of decimal places.

### Define one operation

`def` defines a function. A **parameter** is the name used for an input inside the definition. `return` gives a result back to the caller. Indent the function body by four spaces. Running the definition creates the function; the final line calls it.

```python
def years_gained(earlier, later):
    return later - earlier

print(years_gained(68.0, 74.0))
```

The result is 6.0 years for fictional Northland. The input order matters: earlier first, later second. The parameter names are local to the function, meaning they belong to that calculation rather than replacing unrelated notebook variables.

### Separate a result from a display

Calling the function returns a value you can store or reuse. The function does not print anything by itself. Use a named argument to make input meanings explicit.

```python
northland_gain = years_gained(earlier=68.0, later=74.0)
print(northland_gain)
print(northland_gain / 2)
```

The stored gain is six years and half of it is three. A function that only prints a value instead of returning it cannot supply that numeric result to another calculation. Keep the calculation and the display separate when writing a reusable function.

### Define percentage growth

For a positive starting level, percentage growth is 100 times the change divided by the starting level. We add a short **docstring**, a sentence inside triple quotation marks describing the function. First test the rule on an invented increase from 100 to 110.

```python
def growth_percent(earlier, later):
    """Percentage change relative to a positive starting level."""
    return 100 * (later - earlier) / earlier

print(growth_percent(100, 110))
```

The answer is 10.0 percent. Dividing by the ending level would answer a different question. A zero starting value makes this formula undefined; negative starting values need a different interpretation. This function expects you to check those assumptions.

### Apply it to a real period

**The running example.** The real observations below come from the course's [historical Gapminder teaching extract](https://github.com/ningzhanglab/Python-for-Beginners/blob/main/data/README.md), not the fictional examples. Canada in 2007 has life expectancy of 80.653 years and GDP per person of 36,319.23501 inflation-adjusted US dollars in this extract. GDP per person measures average economic production, not each person's earnings. These are historical observations, not today's statistics or evidence of causation.

Canada's GDP per person was 33,328.96507 in 2002 and 36,319.23501 in 2007. Call the same function with those two inputs.

```python
canada_growth = growth_percent(33328.96507, 36319.23501)
print(round(canada_growth, 2))
```

The increase is about 8.97 percent **over five years**. It is not an annual growth rate. Differences in measurement and other country characteristics still matter when comparing this with life expectancy.

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### An argument is missing

The function requires both an earlier and a later value.

```python
try:
    years_gained(68.0)
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

The error names the missing parameter, `later`. Supply both inputs in the documented order, or use their parameter names.

### A denominator is zero

The arithmetic formula cannot define percentage growth from a zero starting level.

```python
try:
    growth_percent(0, 110)
except ZeroDivisionError as error:
    print(f"{type(error).__name__}: {error}")
```

Do not replace zero with a tiny number just to suppress this error. Check whether the question should use an absolute change instead, or exclude the case with an explicit justification.
