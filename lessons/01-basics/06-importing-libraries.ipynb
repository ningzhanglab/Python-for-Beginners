## Why it matters

You do not need to write every calculation from scratch. A **module** is a unit of Python code you can import; a **library** is a collection of reusable code, often containing several modules. The **standard library** ships with Python. Other libraries, such as pandas for tables, are installed as packages in the computing environment. This lesson uses only what the course environment and ordinary Colab sessions already provide.

## Worked example

### Import a standard module

`import math` makes the math module available. The dot in `math.sqrt` looks up its square-root function. Test it with a made-up variance of 25 square years; the square root is a standard deviation in years.

```python
import math
fictional_variance = 25
print(math.sqrt(fictional_variance))
```

The answer is 5.0 years. `math` is part of Python itself, so no separate package installation is needed. A variance and its square root have different units.

### Use a short alias

`as` gives an imported module a shorter name. pandas is conventionally imported as `pd`. The alias is a local convenience, not a different library. This small table contains made-up numbers for fictional countries.

```python
import pandas as pd
small = pd.DataFrame({"country": ["Northland", "Eastria"], "life_exp": [74.0, 78.0]})
print(small.shape)
small
```

A **DataFrame** is a table with named columns. `.shape` reports rows and columns: (2, 2). The names before colons identify columns; the lists after them give the values. Lesson 2.1 explains this structure slowly and loads it from a file.

### Prepare the real country panel

Run this prepared setup once in a fresh session. It uses the course CSV in a local checkout, or downloads its fixed, checksum-verified public source in Colab. A checksum is a fingerprint of a file's contents; a mismatch stops the download from being used. You do not need to retype the setup. It does not use your Google Drive or any private data.

```python
# Prepared setup: use the course CSV locally; fetch its pinned source in Colab.
from pathlib import Path
import hashlib
import io
from urllib.request import urlopen
import pandas as pd

data_file = Path("data/gapminder.csv")
if not data_file.exists():
    source = "https://raw.githubusercontent.com/jennybc/gapminder/391b5a40d3574c3b89d889dd7d72f7f3fdb4a70c/inst/extdata/gapminder.tsv"
    with urlopen(source, timeout=30) as response:
        raw = response.read()
    expected = "f74275b4a2f4a7c8e3a09a752f9d6e43d1759d78d5cbfdebaaa3e1a6efc68f78"
    if hashlib.sha256(raw).hexdigest() != expected:
        raise ValueError("The data download changed. Please report this to the course maintainer.")
    downloaded = pd.read_csv(io.BytesIO(raw), sep="\t", dtype=str)
    downloaded = downloaded.rename(columns={"continent": "region", "lifeExp": "life_exp", "pop": "population", "gdpPercap": "gdp_pc"})
    data_file.parent.mkdir(parents=True, exist_ok=True)
    downloaded.to_csv(data_file, index=False, lineterminator="\n")
print("Course data ready: data/gapminder.csv")
```

Look for **Course data ready**. Colab working files are temporary; rerun this cell after a runtime reset. The data cover 142 countries at five-year intervals from 1952 to 2007. Each row is one country in one year.

### Use the imported library on the real file

### Identify the environment

A **version** identifies a release of a library. `pd.__version__` displays pandas' release label. `sys.version_info` describes the Python interpreter, the program that reads Python instructions. These prepared commands help you report your environment when asking for help.

```python
import sys
print("pandas:", pd.__version__)
print("Python:", sys.version_info.major, sys.version_info.minor)
```

## When it goes wrong

The following examples deliberately trigger specific errors. The prepared `try`/`except` wrapper catches that named error and prints its message so the notebook can continue. It is for these demonstrations; an unexpected error elsewhere should still stop execution. You can read the message without memorizing the wrapper.

### A module name is misspelled

This name deliberately does not refer to an installed module.

```python
try:
    import pandas_typo_course_example
except ModuleNotFoundError as error:
    print(f"{type(error).__name__}: {error}")
```

Check spelling first. A ModuleNotFoundError can also mean a package is absent from the active environment. For this course, return to the documented environment or a fresh Colab runtime before adding unrelated installation commands.

### A function name is misspelled

A module can be imported successfully while a requested name inside it is absent.

```python
try:
    math.squareroot(25)
except AttributeError as error:
    print(f"{type(error).__name__}: {error}")
```

AttributeError means the named object does not have that attribute or operation. The function is `math.sqrt`, not `math.squareroot`. Use the library's official documentation to check spelling and arguments.
